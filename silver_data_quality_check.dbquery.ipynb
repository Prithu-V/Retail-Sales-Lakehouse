```sql
select * from retail_lakehouse.retail.silver_retail_transactions
limit 10;

-- ########################################################

select count(*) as total_rows,
count(distinct invoice_id) as distinct_invoices,
count(distinct stock_code) as distinct_stocks,
MIN(invoice_timestamp) AS min_date,
MAX(invoice_timestamp) AS max_date
FROM retail_lakehouse.retail.silver_retail_transactions;

-- ########################################################

SELECT
    SUM(CASE WHEN invoice_id IS NULL THEN 1 ELSE 0 END) AS null_invoice_ids,
    SUM(CASE WHEN stock_code IS NULL THEN 1 ELSE 0 END) AS null_stock_codes,
    SUM(CASE WHEN quantity IS NULL THEN 1 ELSE 0 END) AS null_quantities,
    SUM(CASE WHEN unit_price IS NULL THEN 1 ELSE 0 END) AS null_unit_prices,
    SUM(CASE WHEN customer_id IS NULL THEN 1 ELSE 0 END) AS missing_customer_ids,
    SUM(CASE WHEN invoice_date IS NULL THEN 1 ELSE 0 END) AS missing_invoice_dates,
    SUM(CASE WHEN line_amount IS NULL THEN 1 ELSE 0 END) AS null_line_amounts
FROM retail_lakehouse.retail.silver_retail_transactions;

-- ########################################################

SELECT
    SUM(CASE WHEN quantity < 0 THEN 1 ELSE 0 END)
        AS negative_quantity_rows,
    SUM(CASE WHEN unit_price < 0 THEN 1 ELSE 0 END)
        AS negative_price_rows,
    SUM(CASE WHEN unit_price = 0 THEN 1 ELSE 0 END)
        AS zero_price_rows,
    MIN(quantity) AS minimum_quantity,
    MIN(unit_price) AS minimum_unit_price,
    MIN(line_amount) AS minimum_line_amount
FROM retail_lakehouse.retail.silver_retail_transactions;

-- ########################################################

SELECT
    transaction_type,
    COUNT(*) AS row_count,
    SUM(line_amount) AS total_line_amount
FROM retail_lakehouse.retail.silver_retail_transactions
GROUP BY transaction_type
ORDER BY row_count DESC;

create schema if not exists retail_lakehouse.gold
```
