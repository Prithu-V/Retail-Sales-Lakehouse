```sql
create or replace table retail_lakehouse.gold.gold_country_performance as
with country_performance as (
    select 
    country,
    sum(line_amount) as total_revenue,
    sum(quantity) as unit_sold,
    count(distinct invoice_id) as order_count,
    count(distinct customer_id) as customer_count,
    round((sum(line_amount) / count(distinct invoice_id)),2) as avg_order_value
    from retail_lakehouse.retail.silver_retail_transactions
    where transaction_type = 'SALE'
    group by country
)

select * from country_performance;
select * from retail_lakehouse.gold.gold_country_performance;

select sum(total_revenue) from retail_lakehouse.gold.gold_country_performance
```
