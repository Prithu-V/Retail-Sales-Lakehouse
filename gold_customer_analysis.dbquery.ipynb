```sql
create or replace table retail_lakehouse.gold.gold_customer_analysis as
with customer_analysis as(
    select 
    customer_id,
    sum(line_amount) as total_spend,
    sum(quantity) as total_units_purchased,
    count(distinct invoice_id) as order_count,
    round(sum(line_amount)/count(distinct invoice_id),2) as avg_order_value,
    min(invoice_date) as first_purchase_date,
    max(invoice_date) as last_purchase_date
    from retail_lakehouse.retail.silver_retail_transactions
    where transaction_type = 'SALE' and customer_id is not null
    group by customer_id
)

select * from customer_analysis;

select sum(total_spend) from retail_lakehouse.gold.gold_customer_analysis;
```
