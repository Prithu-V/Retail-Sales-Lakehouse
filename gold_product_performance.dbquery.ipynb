```sql
create or replace table retail_lakehouse.gold.gold_product_performance as
with product_performance as(
    select 
    stock_code,
    description,
    sum(line_amount) as total_revenue,
    sum(quantity) as unit_sold,
    count(distinct invoice_id) as invoice_count,
    avg(unit_price) as avg_selling_price
    from retail_lakehouse.retail.silver_retail_transactions
    WHERE transaction_type = 'SALE'
    group by stock_code,
    description
)

select * from product_performance;

SELECT *
FROM retail_lakehouse.gold.gold_product_performance
ORDER BY total_revenue DESC
LIMIT 10;

SELECT SUM(total_revenue) AS gold_product_revenue
FROM retail_lakehouse.gold.gold_product_performance;
```
